```python
import pandas as pd
```

```python
df = pd.read_csv('Data/data.csv',
                 header=0,
                 encoding='unicode_escape',
                 dtype={'CustomerID': str,
                        'InvoiceID': str},
                 parse_dates=['InvoiceDate'],
                 infer_datetime_format=True)
df.head()
```

# Reducing the size of the data set

```python
sampleData = df.sample(frac=0.3)

sampleData[sampleData["CustomerID"] != 'NaN']
```

```python
mbAnalysis = sampleData[['CustomerID', 'InvoiceDate', 'Description']]
mbAnalysis.head()
```

```python
mbAnalysis['single_transaction'] = df['CustomerID'].astype(
    str) + '_' + df['InvoiceDate'].astype(str).str.split(" ").str[0]

mbAnalysis.head()
```

```python
df2 = pd.crosstab(mbAnalysis['single_transaction'],
                  mbAnalysis['Description'])
df2.head()
```

```python
def encode(item_freq):
    res = 0
    if item_freq > 0:
        res = 1
    return res


basket_input = df2.applymap(encode)
```

```python
sample_train =  basket_input.sample(frac=0.35,random_state=42)
```

```python
from mlxtend.frequent_patterns import apriori
from mlxtend.frequent_patterns import association_rules

frequent_itemsets = apriori(sample_train,
                            min_support=0.005,
                            use_colnames=True)

rules = association_rules(frequent_itemsets, metric="lift")

rules
```

```python
rules.sort_values(["support", "confidence","lift"],axis = 0, ascending = False).head(8)
```
